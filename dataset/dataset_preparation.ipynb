```python
from sklearn.model_selection import train_test_split
from pathlib import Path
import shutil
```

```python
dataset = Path("Plant_leave_diseases_dataset_without_augmentation")
output_path = Path("PlantVillage_split")
```

```python
images = {}

for class_dir in dataset.iterdir():
    if class_dir.is_dir():
        images[class_dir.name] = []
        for image_path in class_dir.iterdir():
            if image_path.is_file():
                images[class_dir.name].append(image_path)
```

```python
train_images = {}
val_images = {}

for class_name, image_paths in images.items():
    train, val = train_test_split(image_paths, test_size=0.20, shuffle=True, random_state=42)

    train_images[class_name] = train
    val_images[class_name] = val
```

```python
print(len(train_images["Apple___Apple_scab"]))
print(len(val_images["Apple___Apple_scab"]))
```

```python
test_images = {}
val_images_new = {}

for class_name, image_paths in val_images.items():
    test, val = train_test_split(image_paths, test_size=0.50, shuffle=True, random_state=42)

    test_images[class_name] = test
    val_images_new[class_name] = val
```

```python
print(f"Train: {len(train_images["Apple___Apple_scab"])}")
print(f"Val: {len(val_images_new["Apple___Apple_scab"])}")
print(f"Test: {len(test_images["Apple___Apple_scab"])}")
print(f"Total: {len(images["Apple___Apple_scab"])}")
```

```python
train_path = output_path / "train"
val_path = output_path / "val"
test_path = output_path / "test"

train_path.mkdir(parents=True, exist_ok=True)
val_path.mkdir(parents=True, exist_ok=True)
test_path.mkdir(parents=True, exist_ok=True)
```

```python
def copy_dataset(images_dict, output_dir):
    for class_name, image_paths in images_dict.items():
        class_dir = output_dir / class_name
        class_dir.mkdir(parents=True, exist_ok=True)

        for image_path in image_paths:
            shutil.copy2(image_path, class_dir / image_path.name)
```

```python
copy_dataset(train_images, train_path)
copy_dataset(val_images_new, val_path)
copy_dataset(test_images, test_path)
```
